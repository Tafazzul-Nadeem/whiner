```python
import json

# Load JSON
with open("data/cleaned_mails.json", "r") as f:
    data = json.load(f)

# Access and print email contents
for entry in data["mail"]:
    print("-" * 120)
    print("Subject:", entry["subject"])
    print("Date:", entry["date"])
    print("Body:", entry["body"])
    print("Attachment:", entry["attachment"])
    print("-" * 120)
```
